```python
import pandas as pd
from display import Display

# Load initial and main data
before = pd.read_csv('data/births.csv')

births = pd.read_csv('data/births.csv')

# Create decade column to group by and makes using pivot table easier
births['decade'] = 10 * (births['year'] // 10) # example: 1969 // 10 = 196 * 10 = 1960

Display(before=before.head(), births=births.head())
```

```python
pivot_table = births.pivot_table(
    'births', # df
    index='decade', # key to group on index
    columns='gender', # key to group on columns
    aggfunc='sum' # aggregation in this case, sum correspondings columns
)


Display(births=births.head(),pivot_table=pivot_table)
```

```python
import matplotlib.pyplot as plt

pivot_table.plot()
plt.ylabel('total births per year')
```
